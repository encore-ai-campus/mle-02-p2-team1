```python
import sys
import importlib.util

print("현재 파이썬:", sys.executable)

if importlib.util.find_spec("neo4j_graphrag") is not None:
    print("neo4j_graphrag 패키지가 설치되어 있어요.")
else:
    print("neo4j_graphrag 패키지가 설치되어 있지 않아요.")
```

```python
from neo4j_graphrag.components.text_splitters.fixed_size_splitter import FixedSizeSplitter

CHUNK_SIZE=1200
CHUNK_OVERLAP=150

# 청킹 도구 설정
splitter = FixedSizeSplitter(
    chunk_size=CHUNK_SIZE,
    chunk_overlap=CHUNK_OVERLAP,
)

if "news" not in globals():
    import json
    from pathlib import Path

    project_root = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "data/collected/news.jsonl").is_file())
    with (project_root / "data/collected/news.jsonl").open(encoding="utf-8") as file:
        news = json.loads(next(line for line in file if line.strip()))

# 현재 불러온 뉴스의 본문을 청킹
result = await splitter.run(text=news["content"])

print("기사 ID:", news["document_id"])
print("기사 본문 길이:", len(news["content"]), "글자")
print("청크 개수:", len(result.chunks))

for chunk in result.chunks:
    print(f"청크 {chunk.index}: {len(chunk.text)}글자")
```

```python
for chunk in result.chunks:
    print(f"\n===== 청크 {chunk.index} =====")
    print(chunk.text)
```

```python
print("===== 청크 0의 마지막 200자 =====")
print(result.chunks[0].text[-200:])

print("\n===== 청크 1의 첫 200자 =====")
print(result.chunks[1].text[:200])
```

```python
chunk_records = []

for chunk in result.chunks:
    record = {
        "document_id": news["document_id"],
        "chunk_index": chunk.index,
        "text": chunk.text,
    }
    chunk_records.append(record)

print("만든 청크 기록 수:", len(chunk_records))
print("첫 번째 청크의 기사 ID:", chunk_records[0]["document_id"])
print("첫 번째 청크 번호:", chunk_records[0]["chunk_index"])
print("첫 번째 청크 앞부분:", chunk_records[0]["text"][:100])
```

```python
for record in chunk_records:
    record["title"] = news["title"]
    record["original_url"] = news["original_url"]
    record["published_at"] = news["published_at"]

print("기사 ID:", chunk_records[0]["document_id"])
print("기사 제목:", chunk_records[0]["title"])
print("원문 주소:", chunk_records[0]["original_url"])
print("발행일:", chunk_records[0]["published_at"])
```

```python
for record in chunk_records:
    record["chunk_id"] = (
        f'{record["document_id"]}-{record["chunk_index"]}'
    )

for record in chunk_records:
    print(
        "청크 ID:", record["chunk_id"],
        "| 본문 길이:", len(record["text"]), "글자"
    )
```

```python
import json
import hashlib
from pathlib import Path

news_path = Path("/Users/anna/bulsae/data/collected/news.jsonl")

all_chunk_records = []
document_count = 0

with news_path.open("r", encoding="utf-8") as file:
    for line in file:
        if not line.strip():
            continue

        article = json.loads(line)
        document_count += 1

        # 원본 본문의 변경 여부를 구분할 수 있는 값
        content_hash = hashlib.sha256(
            article["content"].encode("utf-8")
        ).hexdigest()[:12]

        # 앞에서 만든 splitter를 그대로 사용
        result = await splitter.run(text=article["content"])

        for chunk in result.chunks:
            record = {
                "chunk_id": f'{article["document_id"]}-{content_hash}-{chunk.index}',
                "document_id": article["document_id"],
                "chunk_index": chunk.index,
                "text": chunk.text,
                "title": article["title"],
                "original_url": article["original_url"],
                "published_at": article["published_at"],
                "company_ids": article["company_ids"],
            }

            all_chunk_records.append(record)

print("처리한 기사 수:", document_count)
print("만든 청크 수:", len(all_chunk_records))
print("첫 번째 청크 ID:", all_chunk_records[0]["chunk_id"])
```

```python
#청크 아이디 중복확인 

chunk_ids = [record["chunk_id"] for record in all_chunk_records]

print("전체 청크 ID 수:", len(chunk_ids))
print("서로 다른 청크 ID 수:", len(set(chunk_ids)))
print("중복된 청크 ID 수:", len(chunk_ids) - len(set(chunk_ids)))
```

```python
import hashlib
import json
from importlib.metadata import version
from pathlib import Path

# 이번에 사용한 청킹 설정
chunk_config = {
    "splitter": "FixedSizeSplitter",
    "package_version": version("neo4j-graphrag"),
    "chunk_size": splitter.chunk_size,
    "chunk_overlap": splitter.chunk_overlap,
    "approximate": splitter.approximate,
}

# 설정을 구분할 짧은 ID 만들기
config_text = json.dumps(chunk_config, sort_keys=True)
config_id = hashlib.sha256(
    config_text.encode("utf-8")
).hexdigest()[:12]

# 원본 기사와 구분되는 새 파일
output_path = Path("/Users/anna/bulsae/data/processed/news_chunks_v1.jsonl")
output_path.parent.mkdir(parents=True, exist_ok=True)

# 기존 파일이 있으면 덮어쓰지 않고 멈춤
with output_path.open("x", encoding="utf-8") as file:
    for record in all_chunk_records:
        saved_record = record.copy()
        saved_record["chunk_id"] = f'{record["chunk_id"]}-{config_id}'
        saved_record["chunk_config"] = chunk_config

        file.write(json.dumps(saved_record, ensure_ascii=False) + "\n")

print("저장 위치:", output_path)
print("저장한 청크 수:", len(all_chunk_records))
print("청킹 설정:", chunk_config)
```
